```python
import numpy as np

# ==========================================
# 1. INDATA OCH MILJÖ
# ==========================================
n_samples = 100 
alpha = 0.0043         # Aluminium
R_ref = 0.164          # ohm vid 20°C (Referensvärde)
T_ref = 20

# Dina specifika mätvärden
V1_nom = 10000 * np.exp(1j * np.deg2rad(0))
V2_nom = 9945 * np.exp(1j * np.deg2rad(-0.25))
I_nom  = 300 * np.exp(1j * np.deg2rad(-30))

# Simulera brus (Nödvändigt för att Least Squares ska kunna skatta parametrar)
noise_level = 0.0001 
Vs = V1_nom * (1 + noise_level * np.random.randn(n_samples))
Vr = V2_nom * (1 + noise_level * np.random.randn(n_samples))
Ir = I_nom  * (1 + noise_level * np.random.randn(n_samples))

# Antag en mycket liten susceptans (B) för kabeln
B_true = 1e-7 
Is = Ir + 1j * (B_true/2) * (Vs + Vr) 

# ==========================================
# 2. STEG 1: SKATTA K1 OCH B [2, 3]
# ==========================================
Y1 = []
A1 = []
for k in range(n_samples):
    Y1.append(Is[k].real)
    A1.append([Ir[k].real, -Ir[k].imag, -0.5*(Vs[k].imag + Vr[k].imag)])
    Y1.append(Is[k].imag)
    A1.append([Ir[k].imag, Ir[k].real, 0.5*(Vs[k].real + Vr[k].real)])

beta1, _, _, _ = np.linalg.lstsq(A1, Y1, rcond=None)
reK1, imK1, B_est = beta1
K1_complex = reK1 + 1j * imK1

# ==========================================
# 3. STEG 2: SKATTA K4, R OCH X [4, 5]
# ==========================================
Y2 = []
A2 = []
for k in range(n_samples):
    I_eq = (1j * (B_est / 2) * Vr[k]) + Ir[k]
    
    Y2.append(Vs[k].real)
    A2.append([Vr[k].real, -Vr[k].imag, I_eq.real, -I_eq.imag])
    Y2.append(Vs[k].imag)
    A2.append([Vr[k].imag, Vr[k].real, I_eq.imag, I_eq.real])

beta2, _, _, _ = np.linalg.lstsq(A2, Y2, rcond=None)
reK4, imK4, R_est, X_est = beta2
K4_complex = reK4 + 1j * imK4

# ==========================================
# 4. TEMPERATURBERÄKNING [6]
# ==========================================
T_calc = ((R_est / R_ref) - 1) / alpha + T_ref

print("--- RESULTAT ---")
print(f"Skatta K1: {K1_complex:.6f}")
print(f"Skatta K4: {K4_complex:.6f}")
print(f"Skatta R (Resistans): {R_est:.6f} ohm")
print(f"Skatta X (Reaktans): {X_est:.6f} ohm")
print(f"Beräknad kabeltemperatur: {T_calc:.2f} °C")
```

```python
# Assistant
# Referenstemperatur för resistansberäkning
vid 20°C (Referensvärde)
T_ref = 20

# Definierar nominella värden för spänningar och ström
# V1_nom: Spänning vid sändarsidan (10 kV med 0° fasvinkel)
# V2_nom: Spänning vid mottagarsidan (9.945 kV med -0.25° fasvinkel)
# I_nom: Nominell ström (300 A med -30° fasvinkel)
V1_nom = 10000 * np.exp(1j * np.deg2rad(0))
V2_nom = 9945 * np.exp(1j * np.deg2rad(-0.25))
I_nom  = 300 * np.exp(1j * np.deg2rad(-30))

# Skapar simulerade mätvärden med litet brus för att göra skattningen möjlig
# Brus är nödvändigt för att least squares-algoritmen ska fungera korrekt
noise_level = 0.0001 
Vs = V1_nom * (1 + noise_level * np.random.randn(n_samples))  # Spänning sändarsida
Vr = V2_nom * (1 + noise_level * np.random.randn(n_samples))  # Spänning mottagarsida
Ir = I_nom  * (1 + noise_level * np.random.randn(n_samples))  # Ström mottagarsida

# Definierar ett sant värde för susceptans (B) för kabeln
# Beräknar sändarström (Is) baserat på mottagarström och susceptans
B_true = 1e-7 
Is = Ir + 1j * (B_true/2) * (Vs + Vr)  # Ekvation för sändarström med shuntadmittans

# ==========================================
# 2. STEG 1: SKATTA K1 OCH B [2, 3]
# ==========================================
# Skapar matriser för least squares-skattning av K1 och B
Y1 = []  # Målvärden (sändarström)
A1 = []  # Designmatris
for k in range(n_samples):
    # Lägger till realdel av Is som målvärde
    Y1.append(Is[k].real)
    # Motsvarande rad i designmatrisen för realdelen
    A1.append([Ir[k].real, -Ir[k].imag, -0.5*(Vs[k].imag + Vr[k].imag)])
    
    # Lägger till imaginärdel av Is som målvärde
    Y1.append(Is[k].imag)
    # Motsvarande rad i designmatrisen för imaginärdelen
    A1.append([Ir[k].imag, Ir[k].real, 0.5*(Vs[k].real + Vr[k].real)])

# Utför least squares-skattning för att få K1 och B
beta1, _, _, _ = np.linalg.lstsq(A1, Y1, rcond=None)
reK1, imK1, B_est = beta1  # Extraherar skattade parametrar
K1_complex = reK1 + 1j * imK1  # Skapar komplex K1

# ==========================================
# 3. STEG 2: SKATTA K4, R OCH X [4, 5]
# ==========================================
# Skapar matriser för least squares-skattning av K4, R och X
Y2 = []  # Målvärden (sändarspänning)
A2 = []  # Designmatris
for k in range(n_samples):
    # Beräknar ekvivalent ström med den skattade susceptansen
    I_eq = (1j * (B_est / 2) * Vr[k]) + Ir[k]
    
    # Lägger till realdel av Vs som målvärde
    Y2.append(Vs[k].real)
    # Motsvarande rad i designmatrisen för realdelen
    A2.append([Vr[k].real, -Vr[k].imag, I_eq.real, -I_eq.imag])
    
    # Lägger till imaginärdel av Vs som målvärde
    Y2.append(Vs[k].imag)
    # Motsvarande rad i designmatrisen för imaginärdelen
    A2.append([Vr[k].imag, Vr[k].real, I_eq.imag, I_eq.real])

# Utför least squares-skattning för att få K4, R och X
beta2, _, _, _ = np.linalg.lstsq(A2, Y2, rcond=None)
reK4, imK4, R_est, X_est = beta2  # Extraherar skattade parametrar
K4_complex = reK4 + 1j * imK4  # Skapar komplex K4

# ==========================================
# 4. TEMPERATURBERÄKNING [6]
# ==========================================
# Beräknar kabeltemperatur baserat på skattad resistans
# Använder formeln: T = ((R/R_ref) - 1)/alpha + T_ref
T_calc = ((R_est / R_ref) - 1) / alpha + T_ref

# Skriver ut resultaten
print("--- RESULTAT ---")
print(f"Skatta K1: {K1_complex:.6f}")
print(f"Skatta K4: {K4_complex:.6f}")
print(f"Skatta R (Resistans): {R_est:.6f} ohm")
print(f"Skatta X (Reaktans): {X_est:.6f} ohm")
print(f"Beräknad kabeltemperatur: {T_calc:.2f} °C")
```

```python
import numpy as np

# ==========================================
# 1. KALIBRERING & DATA GENERERING (Fas 1)
# ==========================================
n_samples = 1000 
alpha, R_ref, T_ref = 0.0043, 0.164, 20
X_true = 0.38 * 20 # Baserat på 0.38 ohm/km för en 20km ledning [10]

# Skapa varierande last (viktigt för stabilitet!)
current_profile = 300 + 100 * np.sin(np.linspace(0, 2*np.pi, n_samples))
I_nom = current_profile * np.exp(1j * np.deg2rad(-30))

# Generera spänningar utifrån en känd temperatur (t.ex. 50 grader)
T_true = 50
R_true = R_ref * (1 + alpha * (T_true - T_ref))
Z_true = R_true + 1j * X_true

Vr = 9945 * np.exp(1j * np.deg2rad(-0.25))
Vs = Vr + Z_true * I_nom # Förenklad modell för demo
noise = 0.0001
Vs_noisy = Vs * (1 + noise * np.random.randn(n_samples))

# ==========================================
# 2. IMPLEMENTERA X SOM ANKARE (Fas 2)
# ==========================================
# Antag att vi har gjort kalibrering och vet att X_baseline = 7.6 ohm
X_baseline = 7.6 

Y_anchor = []
A_anchor = []

for k in range(n_samples):
    # Vi flyttar den kända reaktansen till Y-vektorn (vänster sida)
    # Ekvation: Vs - j*X_base*Ir = K4*Vr + R*Ir
    V_fixed = Vs_noisy[k] - (1j * X_baseline * I_nom[k])
    
    # Real-del rad
    Y_anchor.append(V_fixed.real)
    A_anchor.append([Vr.real, -Vr.imag, I_nom[k].real]) 
    
    # Imag-del rad
    Y_anchor.append(V_fixed.imag)
    A_anchor.append([Vr.imag, Vr.real, I_nom[k].imag])

# Lös för [Re(K4), Im(K4), R]
beta, _, _, _ = np.linalg.lstsq(A_anchor, Y_anchor, rcond=None)
reK4, imK4, R_estimated = beta

# ==========================================
# 3. RESULTAT
# ==========================================
T_calc = ((R_estimated / R_ref) - 1) / alpha + T_ref

print(f"--- RESULTAT MED X-ANKARE ({X_baseline} ohm) ---")
print(f"Skatta R: {R_estimated:.6f} ohm")
print(f"Skatta K4: {reK4 + 1j*imK4:.6f}")
print(f"Beräknad kabeltemperatur: {T_calc:.2f} °C")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

# ==========================================
# 1. KONFIGURATION OCH PARAMETRAR
# ==========================================
DATA_PATH = r"C:\Users\lundi\OneDrive\Skóli\Svíþjöð\Exjobb\Arbetet\Källor\phasor_dlr-main\phasor_dlr-main\data"
SEND_FILE = os.path.join(DATA_PATH, "send-station.xlsx")
REC_FILE = os.path.join(DATA_PATH, "rec-station.xlsx")

# Kabelparametrar (Aluminium 15.313 km) [3]
LÄNGD = 15.313
X_BASELINE = 0.38 * LÄNGD   # 0.38 ohm/km baslinje [4, 5]
R_REF = 1.98                # AC-resistans vid 20°C [3]
ALPHA = 0.00403             # Temperaturkoefficient Al [6]

# Algoritm-inställningar
WINDOW_SIZE = 3600          # 1 timme fönster för stabilitet [7]
STEP_SIZE = 300             # Beräkna var 5:e minut

def load_pmu_data(file_path, key_sheet, data_sheet):
    """Läser GUID-nycklar och transformerar mätdata till kolumner."""
    # 1. Läs nycklar (A=Parameter, B=ID)
    keys = pd.read_excel(file_path, sheet_name=key_sheet, header=None, names=['Parameter', 'ID'])
    keys = keys.dropna(subset=['Parameter', 'ID'])
    id_mapping = dict(zip(keys['ID'], keys['Parameter']))
    
    # 2. Läs mätdata (A=Timestamp, B=ID, C=Value)
    data = pd.read_excel(file_path, sheet_name=data_sheet, names=['Timestamp', 'ID', 'Value'])
    
    # 3. Filtrera och pivotera data
    data = data[data['ID'].isin(id_mapping.keys())].copy()
    data['Parameter'] = data['ID'].map(id_mapping)
    
    # Skapa brett format: Rader=Timestamp, Kolumner=Parameter
    df_wide = data.pivot(index='Timestamp', columns='Parameter', values='Value')
    return df_wide

# ==========================================
# 2. INLÄSNING OCH SYNKRONISERING
# ==========================================
print("Läser och transformerar Send-station (K-nyckel)...")
df_s = load_pmu_data(SEND_FILE, 'K-nyckel', 'Station_K_3x15000')

print("Läser och transformerar Rec-station (F-nyckel)...")
df_r = load_pmu_data(REC_FILE, 'F-nyckel', 'Station_F_3x15000')

# Merge på Timestamp för att säkerställa synkronisering [8]
df = pd.merge(df_s, df_r, left_index=True, right_index=True, suffixes=('_S', '_R')).dropna()

def to_complex(mag, ang_deg):
    return mag * np.exp(1j * np.deg2rad(ang_deg))

# Skapa komplexa fasorer för Fas L1
Vs = to_complex(df['UL1MAG_S'], df['UL1ANG_S']).values
Vr = to_complex(df['UL1MAG_R'], df['UL1ANG_R']).values
Is = to_complex(df['IL1MAG_S'], df['IL1ANG_S']).values
Ir = to_complex(df['IL1MAG_R'], df['IL1ANG_R']).values

# ==========================================
# 3. SINGH ET AL. ALGORITM (X-ANKARE) [1, 9]
# ==========================================
results_T = []
timestamps = []

print(f"Startar analys på {len(df)} synkroniserade rader...")

for i in range(0, len(df) - WINDOW_SIZE, STEP_SIZE):
    v_s_win, v_r_win = Vs[i:i+WINDOW_SIZE], Vr[i:i+WINDOW_SIZE]
    i_s_win, i_r_win = Is[i:i+WINDOW_SIZE], Ir[i:i+WINDOW_SIZE]
    
    # --- STEG 1: Skatta B (Susceptans) och K1 [1] ---
    Y1, A1 = [], []
    for k in range(WINDOW_SIZE):
        Y1.extend([i_s_win[k].real, i_s_win[k].imag])
        # [reK1, -imK1, B_term]
        A1.append([i_r_win[k].real, -i_r_win[k].imag, -0.5*(v_s_win[k].imag + v_r_win[k].imag)])
        A1.append([i_r_win[k].imag, i_r_win[k].real, 0.5*(v_s_win[k].real + v_r_win[k].real)])
    
    beta1, _, _, _ = np.linalg.lstsq(A1, Y1, rcond=None)
    B_est = beta1[10]
    
    # --- STEG 2: Skatta R och K4 med låst X (Ankare) [5, 11] ---
    Y2, A2 = [], []
    for k in range(WINDOW_SIZE):
        I_eq = (1j * (B_est / 2) * v_r_win[k]) + i_r_win[k]
        # Flytta känt X till vänster sida för stabilitet
        V_fixed = v_s_win[k] - (1j * X_BASELINE * I_eq)
        
        Y2.extend([V_fixed.real, V_fixed.imag])
        A2.append([v_r_win[k].real, -v_r_win[k].imag, I_eq.real])
        A2.append([v_r_win[k].imag, v_r_win[k].real, I_eq.imag])
        
    beta2, _, _, _ = np.linalg.lstsq(A2, Y2, rcond=None)
    R_est = beta2[10]
    
    # Beräkna temperatur [12]
    temp = ((R_est / R_REF) - 1) / ALPHA + 20
    results_T.append(temp)
    timestamps.append(df.index[i + WINDOW_SIZE])

# ==========================================
# 4. VISUALISERING
# ==========================================
plt.figure(figsize=(12, 5))
plt.plot(timestamps, results_T, label='Skattad ledartemperatur L1')
plt.axhline(y=90, color='r', linestyle='--', label='XLPE Gräns (90°C)')
plt.title('PMU-baserad Temperaturövervakning (Singh et al. + X-ankare)')
plt.ylabel('Temperatur (°C)')
plt.grid(True)
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
