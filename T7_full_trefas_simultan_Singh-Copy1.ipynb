```python
# Om något bibliotek saknas:
# !pip install pandas numpy matplotlib tqdm openpyxl

import os
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tqdm.notebook import tqdm

pd.set_option("display.max_columns", None)
pd.options.mode.chained_assignment = None

# ============================================================
# FILVÄGAR OCH BLADNAMN
# ============================================================

send_file = Path(r"C:\Users\lundi\OneDrive\Skóli\Svíþjóð\Exjobb\Arbetet\Källor\phasor_dlr-main\phasor_dlr-main\data\send-station.xlsx")
rec_file  = Path(r"C:\Users\lundi\OneDrive\Skóli\Svíþjóð\Exjobb\Arbetet\Källor\phasor_dlr-main\phasor_dlr-main\data\rec-station.xlsx")

keyS_sheet  = "K-nyckel"
dataS_sheet = "Station_K_3x15000"

keyR_sheet  = "F-nyckel"
dataR_sheet = "Station_F_3x15000"

print("send finns:", send_file.exists())
print("rec finns :", rec_file.exists())

# ============================================================
# SPARA
# ============================================================

output_dir = send_file.parent / "test7_singh_3ph_rowwise_pairing"
output_dir.mkdir(parents=True, exist_ok=True)

print("Output-mapp:", output_dir)

# ============================================================
# HJÄLPFUNKTIONER
# ============================================================

def wrap_angle_deg(angle_deg):
    return (angle_deg + 180.0) % 360.0 - 180.0


def read_key_table_safe(filename, sheetname):
    """
    Läser nyckeltabell:
        Parameter | ID
    """
    T = pd.read_excel(filename, sheet_name=sheetname, header=None)
    T = T.iloc[:, :2].copy()
    T.columns = ["Parameter", "ID"]

    T = T.dropna(how="all").copy()
    T["Parameter"] = T["Parameter"].astype(str).str.strip()
    T["ID"] = T["ID"].astype(str).str.strip()

    # Ta bort eventuell header-rad
    T = T[T["Parameter"].str.lower() != "parameter"].copy()
    T = T[T["ID"].str.lower() != "id"].copy()

    return T


def read_data_table_safe(filename, sheetname):
    """
    Läser datatabell:
        Timestamp | ID | Value

    Behåller radordning och sparar Excel-rad.
    """
    D = pd.read_excel(filename, sheet_name=sheetname, header=None)
    D = D.iloc[:, :3].copy()
    D.columns = ["Timestamp", "ID", "Value"]

    D["ExcelRow"] = np.arange(1, len(D) + 1)

    D["Timestamp"] = pd.to_datetime(D["Timestamp"], errors="coerce")
    D["ID"] = D["ID"].astype(str).str.strip()

    D["Value"] = (
        D["Value"]
        .astype(str)
        .str.strip()
        .str.replace(",", ".", regex=False)
    )
    D["Value"] = pd.to_numeric(D["Value"], errors="coerce")

    # Tar bort eventuell header-rad och rader utan timestamp
    D = D.dropna(subset=["Timestamp"]).copy()

    # DataRow = första riktiga datarad blir 0
    D["DataRow"] = np.arange(len(D))

    return D


def map_ids_to_parameters(data_df, key_df, station_name):
    """
    Kopplar ID till parameter men behåller radordning.
    """
    mapped = data_df.merge(key_df, on="ID", how="left", sort=False)

    missing = mapped[mapped["Parameter"].isna()].copy()
    if len(missing) > 0:
        print(f"\nVARNING: {station_name} har ID som saknas i nyckeltabellen.")
        display(missing[["ExcelRow", "Timestamp", "ID", "Value"]].head(20))
        raise ValueError(f"{station_name}: vissa ID kunde inte mappas till Parameter.")

    mapped = mapped.sort_values("DataRow").reset_index(drop=True)
    return mapped


def build_joined_dataframe_rowwise(send_file, rec_file):
    """
    Bygger df_joined med radvis parning:

        send DataRow 0 paras med rec DataRow 0
        send DataRow 1 paras med rec DataRow 1
        osv.

    Varje 12 rader bildar ett komplett sample:
        UL1ANG, UL1MAG, UL2ANG, UL2MAG, UL3ANG, UL3MAG,
        IL1ANG, IL1MAG, IL2ANG, IL2MAG, IL3ANG, IL3MAG
    """

    expected_params = [
        "UL1ANG", "UL1MAG",
        "UL2ANG", "UL2MAG",
        "UL3ANG", "UL3MAG",
        "IL1ANG", "IL1MAG",
        "IL2ANG", "IL2MAG",
        "IL3ANG", "IL3MAG",
    ]

    n_params_per_sample = len(expected_params)

    # Läs nycklar
    keyS = read_key_table_safe(send_file, keyS_sheet)
    keyR = read_key_table_safe(rec_file, keyR_sheet)

    # Läs data
    send_raw = read_data_table_safe(send_file, dataS_sheet)
    rec_raw  = read_data_table_safe(rec_file, dataR_sheet)

    print("\nRådata:")
    print("send rader:", len(send_raw))
    print("rec  rader:", len(rec_raw))
    print("send tidsintervall:", send_raw["Timestamp"].min(), "->", send_raw["Timestamp"].max())
    print("rec  tidsintervall:", rec_raw["Timestamp"].min(), "->", rec_raw["Timestamp"].max())

    # Mappa ID till Parameter
    send_mapped = map_ids_to_parameters(send_raw, keyS, "send")
    rec_mapped  = map_ids_to_parameters(rec_raw,  keyR, "receive")

    # Kontroll 1: samma antal rader
    if len(send_mapped) != len(rec_mapped):
        raise ValueError(
            f"Send och receive har olika antal datarader: "
            f"send={len(send_mapped)}, rec={len(rec_mapped)}"
        )

    # Kontroll 2: samma Parameter på samma rad
    same_param = (
        send_mapped["Parameter"].reset_index(drop=True)
        ==
        rec_mapped["Parameter"].reset_index(drop=True)
    )

    if not same_param.all():
        mismatch_idx = np.where(~same_param.to_numpy())[0]

        mismatch = pd.DataFrame({
            "DataRow": mismatch_idx,
            "ExcelRow_S": send_mapped.loc[mismatch_idx, "ExcelRow"].to_numpy(),
            "ExcelRow_R": rec_mapped.loc[mismatch_idx, "ExcelRow"].to_numpy(),
            "Timestamp_S": send_mapped.loc[mismatch_idx, "Timestamp"].to_numpy(),
            "Timestamp_R": rec_mapped.loc[mismatch_idx, "Timestamp"].to_numpy(),
            "Parameter_S": send_mapped.loc[mismatch_idx, "Parameter"].to_numpy(),
            "Parameter_R": rec_mapped.loc[mismatch_idx, "Parameter"].to_numpy(),
            "Value_S": send_mapped.loc[mismatch_idx, "Value"].to_numpy(),
            "Value_R": rec_mapped.loc[mismatch_idx, "Value"].to_numpy(),
        })

        mismatch.to_excel(output_dir / "rowwise_parameter_mismatch.xlsx", index=False)

        print("\nPARAMETER-MISMATCH MELLAN SEND OCH RECEIVE:")
        display(mismatch.head(20))

        raise ValueError(
            "Radvis parning är inte säker eftersom Parameter_S och Parameter_R inte matchar. "
            "Se rowwise_parameter_mismatch.xlsx"
        )

    print("\nOK: send och receive har samma Parameter på varje rad.")

    # Kontroll 3: antal rader multipel av 12
    if len(send_mapped) % n_params_per_sample != 0:
        raise ValueError(
            f"Antal rader är inte multipel av {n_params_per_sample}. "
            f"Antal rader = {len(send_mapped)}"
        )

    n_samples = len(send_mapped) // n_params_per_sample

    print("OK: antal rader är multipel av 12.")
    print("Antal samples:", n_samples)

    # Kontroll av första signalblock
    first_send_block = send_mapped["Parameter"].iloc[:n_params_per_sample].tolist()
    first_rec_block = rec_mapped["Parameter"].iloc[:n_params_per_sample].tolist()

    print("\nFörsta signalblock send:")
    print(first_send_block)

    print("\nFörsta signalblock rec:")
    print(first_rec_block)

    if first_send_block != expected_params:
        print("\nOBS: Första send-blocket skiljer sig från expected_params.")
        print("Det behöver inte vara fel, men kontrollera ordningen.")

    if first_rec_block != expected_params:
        print("\nOBS: Första rec-blocket skiljer sig från expected_params.")
        print("Det behöver inte vara fel, men kontrollera ordningen.")

    # Radvis kontrollfil
    rowwise_check = pd.DataFrame({
        "DataRow": send_mapped["DataRow"],
        "ExcelRow_S": send_mapped["ExcelRow"],
        "ExcelRow_R": rec_mapped["ExcelRow"],
        "Timestamp_S": send_mapped["Timestamp"],
        "Timestamp_R": rec_mapped["Timestamp"],
        "Parameter_S": send_mapped["Parameter"],
        "Parameter_R": rec_mapped["Parameter"],
        "Value_S": send_mapped["Value"],
        "Value_R": rec_mapped["Value"],
    })

    rowwise_check["Timestamp_diff_ms"] = (
        rowwise_check["Timestamp_S"] - rowwise_check["Timestamp_R"]
    ).dt.total_seconds() * 1000.0

    rowwise_check.to_excel(output_dir / "rowwise_pairing_check_long.xlsx", index=False)

    print("\nTimestamp-differens radvis [ms]:")
    print(rowwise_check["Timestamp_diff_ms"].describe())

    # Sample index
    send_mapped["SampleNo"] = send_mapped["DataRow"] // n_params_per_sample
    rec_mapped["SampleNo"]  = rec_mapped["DataRow"]  // n_params_per_sample

    send_mapped["WithinSampleRow"] = send_mapped["DataRow"] % n_params_per_sample
    rec_mapped["WithinSampleRow"]  = rec_mapped["DataRow"]  % n_params_per_sample

    # Kontroll: varje sample ska ha 12 unika parametrar
    bad_send = (
        send_mapped.groupby("SampleNo")["Parameter"].nunique()
        != n_params_per_sample
    )
    bad_rec = (
        rec_mapped.groupby("SampleNo")["Parameter"].nunique()
        != n_params_per_sample
    )

    if bad_send.any():
        bad = bad_send[bad_send].index.tolist()
        raise ValueError(f"Send har sample med fel antal parametrar. Exempel: {bad[:10]}")

    if bad_rec.any():
        bad = bad_rec[bad_rec].index.tolist()
        raise ValueError(f"Rec har sample med fel antal parametrar. Exempel: {bad[:10]}")

    # Pivotera till bred form
    wideS = send_mapped.pivot_table(
        index="SampleNo",
        columns="Parameter",
        values="Value",
        aggfunc="first"
    ).add_suffix("_S")

    wideR = rec_mapped.pivot_table(
        index="SampleNo",
        columns="Parameter",
        values="Value",
        aggfunc="first"
    ).add_suffix("_R")

    # Tider per sample
    times = pd.DataFrame({
        "Timestamp_S": send_mapped.groupby("SampleNo")["Timestamp"].first(),
        "Timestamp_R": rec_mapped.groupby("SampleNo")["Timestamp"].first(),
    })

    times["Timestamp_diff_ms"] = (
        times["Timestamp_S"] - times["Timestamp_R"]
    ).dt.total_seconds() * 1000.0

    # Bygg df_joined
    df_joined = times.join(wideS, how="inner").join(wideR, how="inner")
    df_joined.insert(0, "SampleNo", df_joined.index)

    # Använd send-timestamp som index
    df_joined = df_joined.set_index("Timestamp_S", drop=False)
    df_joined.index.name = "Timestamp"

    required_cols = []
    for ph in ["L1", "L2", "L3"]:
        required_cols += [
            f"U{ph}MAG_S", f"U{ph}ANG_S",
            f"U{ph}MAG_R", f"U{ph}ANG_R",
            f"I{ph}MAG_S", f"I{ph}ANG_S",
            f"I{ph}MAG_R", f"I{ph}ANG_R",
        ]

    missing = [c for c in required_cols if c not in df_joined.columns]
    if missing:
        raise ValueError(f"Saknade kolumner efter pivot: {missing}")

    df_joined.reset_index(drop=True).to_excel(
        output_dir / "df_joined_rowwise.xlsx",
        index=False
    )

    print("\nKLART: df_joined skapad med radvis parning.")
    print("Antal samples:", len(df_joined))
    print("Start:", df_joined["Timestamp_S"].min())
    print("Slut :", df_joined["Timestamp_S"].max())
    print("Max timestamp-diff [ms]:", df_joined["Timestamp_diff_ms"].abs().max())

    return df_joined


# ============================================================
# LS-HJÄLPFUNKTIONER
# ============================================================

def ls_diagnostics(A, Y, beta):
    Y_hat = A @ beta
    resid = Y - Y_hat

    n = len(Y)
    p = A.shape[1]
    dof = max(n - p, 1)

    sigma2 = np.sum(resid**2) / dof
    ATA_inv = np.linalg.pinv(A.T @ A)
    cov_beta = sigma2 * ATA_inv
    se_beta = np.sqrt(np.diag(cov_beta))

    return {
        "residuals_vector": resid,
        "sigma2": sigma2,
        "cov_beta": cov_beta,
        "se_beta": se_beta,
        "dof": dof
    }


def estimate_B_and_K1_3ph(vs_dict, vr_dict, is_dict, ir_dict):
    """
    Full simultan trefas steg 1.

    beta1 = [
        Re(K1_L1), Re(K1_L2), Re(K1_L3),
        Im(K1_L1), Im(K1_L2), Im(K1_L3),
        B_L1,      B_L2,      B_L3
    ]
    """
    phases = ["L1", "L2", "L3"]

    Y_parts = []
    A_rows = []

    for ph_idx, ph in enumerate(phases):
        vs = vs_dict[ph]
        vr = vr_dict[ph]
        is_ = is_dict[ph]
        ir = ir_dict[ph]

        ir_re = np.real(ir)
        ir_im = np.imag(ir)

        v_sum = vs + vr
        v_re = np.real(v_sum)
        v_im = np.imag(v_sum)

        # Realdel
        for k in range(len(vs)):
            row = np.zeros(9)
            row[ph_idx] = ir_re[k]
            row[3 + ph_idx] = -ir_im[k]
            row[6 + ph_idx] = 0.5 * v_re[k]
            A_rows.append(row)
            Y_parts.append(np.real(is_[k]))

        # Imaginärdel
        for k in range(len(vs)):
            row = np.zeros(9)
            row[ph_idx] = ir_im[k]
            row[3 + ph_idx] = ir_re[k]
            row[6 + ph_idx] = 0.5 * v_im[k]
            A_rows.append(row)
            Y_parts.append(np.imag(is_[k]))

    A = np.array(A_rows)
    Y = np.array(Y_parts)

    beta, residuals, rank, s = np.linalg.lstsq(A, Y, rcond=None)
    diag = ls_diagnostics(A, Y, beta)

    return {
        "beta": beta,
        "rank": rank,
        "singular_values": s,
        "A": A,
        "Y": Y,
        "cov_beta": diag["cov_beta"],
        "se_beta": diag["se_beta"],
        "sigma2": diag["sigma2"]
    }


def estimate_Z_and_K4_3ph(vs_dict, vr_dict, ir_dict, B_dict):
    """
    Full simultan trefas steg 2.

    beta2 = [
        Re(K4_L1), Re(K4_L2), Re(K4_L3),
        Im(K4_L1), Im(K4_L2), Im(K4_L3),
        R_L1,      R_L2,      R_L3,
        X_L1,      X_L2,      X_L3
    ]
    """
    phases = ["L1", "L2", "L3"]

    Y_parts = []
    A_rows = []

    Ieq_dict = {}

    for ph in phases:
        Ieq_dict[ph] = ir_dict[ph] + 1j * (B_dict[ph] / 2.0) * vr_dict[ph]

    for ph_idx, ph in enumerate(phases):
        vs = vs_dict[ph]
        vr = vr_dict[ph]
        Ieq = Ieq_dict[ph]

        vr_re = np.real(vr)
        vr_im = np.imag(vr)

        ieq_re = np.real(Ieq)
        ieq_im = np.imag(Ieq)

        # Realdel
        for k in range(len(vs)):
            row = np.zeros(12)
            row[ph_idx] = vr_re[k]
            row[3 + ph_idx] = -vr_im[k]
            row[6 + ph_idx] = ieq_re[k]
            row[9 + ph_idx] = -ieq_im[k]
            A_rows.append(row)
            Y_parts.append(np.real(vs[k]))

        # Imaginärdel
        for k in range(len(vs)):
            row = np.zeros(12)
            row[ph_idx] = vr_im[k]
            row[3 + ph_idx] = vr_re[k]
            row[6 + ph_idx] = ieq_im[k]
            row[9 + ph_idx] = ieq_re[k]
            A_rows.append(row)
            Y_parts.append(np.imag(vs[k]))

    A = np.array(A_rows)
    Y = np.array(Y_parts)

    beta, residuals, rank, s = np.linalg.lstsq(A, Y, rcond=None)
    diag = ls_diagnostics(A, Y, beta)

    return {
        "beta": beta,
        "rank": rank,
        "singular_values": s,
        "A": A,
        "Y": Y,
        "Ieq_dict": Ieq_dict,
        "cov_beta": diag["cov_beta"],
        "se_beta": diag["se_beta"],
        "sigma2": diag["sigma2"]
    }


# ============================================================
# 1. BYGG DATA MED RADVIS PARNING
# ============================================================

df_joined = build_joined_dataframe_rowwise(send_file, rec_file)

# ============================================================
# 2. RENSNING
# ============================================================

required_cols = [
    "UL1MAG_S","UL1ANG_S","UL1MAG_R","UL1ANG_R",
    "IL1MAG_S","IL1ANG_S","IL1MAG_R","IL1ANG_R",
    "UL2MAG_S","UL2ANG_S","UL2MAG_R","UL2ANG_R",
    "IL2MAG_S","IL2ANG_S","IL2MAG_R","IL2ANG_R",
    "UL3MAG_S","UL3ANG_S","UL3MAG_R","UL3ANG_R",
    "IL3MAG_S","IL3ANG_S","IL3MAG_R","IL3ANG_R",
]

for c in required_cols:
    df_joined[c] = pd.to_numeric(df_joined[c], errors="coerce")

df_clean = df_joined.dropna(subset=required_cols).copy()

print("\nRader efter rensning:", len(df_clean))
print("Start:", df_clean.index.min())
print("Slut :", df_clean.index.max())
print("Tidslängd:", df_clean.index.max() - df_clean.index.min())

# ============================================================
# 3. BYGG FASORER - MED INVERTERAD MOTTAGARSTRÖM
# ============================================================

phasors = {}

for ph in ["L1", "L2", "L3"]:
    Vs = df_clean[f"U{ph}MAG_S"] * np.exp(1j * np.deg2rad(df_clean[f"U{ph}ANG_S"]))
    Vr = df_clean[f"U{ph}MAG_R"] * np.exp(1j * np.deg2rad(df_clean[f"U{ph}ANG_R"]))
    Is = df_clean[f"I{ph}MAG_S"] * np.exp(1j * np.deg2rad(df_clean[f"I{ph}ANG_S"]))

    # Samma som tidigare Test 7: inverterad mottagarström
    Ir = -df_clean[f"I{ph}MAG_R"] * np.exp(1j * np.deg2rad(df_clean[f"I{ph}ANG_R"]))

    phasors[ph] = {
        "Vs": Vs,
        "Vr": Vr,
        "Is": Is,
        "Ir": Ir,
    }

print("\nFasorer skapade för:", list(phasors.keys()))
print("Mottagarström Ir har inverterats.")

# ============================================================
# 4. FÖNSTERINSTÄLLNINGAR
# ============================================================

dt = df_clean.index.to_series().diff().dropna().dt.total_seconds()

print("\nTidssteg:")
print(dt.describe())
print("\nVanligaste tidssteg:")
print(dt.value_counts().head(10))

fs = 1.0 / dt.median()
print(f"\nUppskattad samplingsfrekvens fs = {fs:.3f} Hz")

window_seconds = 120
step_seconds = 5

WINDOW = int(round(window_seconds * fs))
STEP   = int(round(step_seconds * fs))

R_ref = 1.2575
alpha = 0.00403

N = len(df_clean)

print("\nN =", N)
print("WINDOW =", WINDOW)
print("STEP =", STEP)

if N <= WINDOW:
    raise ValueError(f"För få datapunkter. N={N}, WINDOW={WINDOW}")

idx = np.arange(0, N - WINDOW + 1, STEP)
print("Antal glidfönster =", len(idx))

# ============================================================
# 5. KÖR FULL TREFAS SIMULTAN SINGH
# ============================================================

rows_3ph = []

for i in tqdm(idx, desc="3-ph Singh simultaneous rowwise"):
    i0 = i
    i1 = i + WINDOW

    vs_dict = {}
    vr_dict = {}
    is_dict = {}
    ir_dict = {}

    for ph in ["L1", "L2", "L3"]:
        vs_dict[ph] = phasors[ph]["Vs"].iloc[i0:i1].to_numpy()
        vr_dict[ph] = phasors[ph]["Vr"].iloc[i0:i1].to_numpy()
        is_dict[ph] = phasors[ph]["Is"].iloc[i0:i1].to_numpy()
        ir_dict[ph] = phasors[ph]["Ir"].iloc[i0:i1].to_numpy()

    try:
        # Steg 1
        res1 = estimate_B_and_K1_3ph(vs_dict, vr_dict, is_dict, ir_dict)
        beta1 = res1["beta"]

        K1 = {
            "L1": beta1[0] + 1j * beta1[3],
            "L2": beta1[1] + 1j * beta1[4],
            "L3": beta1[2] + 1j * beta1[5],
        }

        B = {
            "L1": beta1[6],
            "L2": beta1[7],
            "L3": beta1[8],
        }

        # Steg 2
        res2 = estimate_Z_and_K4_3ph(vs_dict, vr_dict, ir_dict, B)
        beta2 = res2["beta"]

        K4 = {
            "L1": beta2[0] + 1j * beta2[3],
            "L2": beta2[1] + 1j * beta2[4],
            "L3": beta2[2] + 1j * beta2[5],
        }

        R = {
            "L1": beta2[6],
            "L2": beta2[7],
            "L3": beta2[8],
        }

        X = {
            "L1": beta2[9],
            "L2": beta2[10],
            "L3": beta2[11],
        }

    except np.linalg.LinAlgError:
        continue

    center_ts = df_clean.index[i0 + WINDOW // 2]

    for ph_idx, ph in enumerate(["L1", "L2", "L3"]):
        se_R = res2["se_beta"][6 + ph_idx]
        se_X = res2["se_beta"][9 + ph_idx]

        T_est = ((R[ph] / R_ref) - 1.0) / alpha + 20.0

        dT_dR = 1.0 / (alpha * R_ref)
        u_T = abs(dT_dR) * se_R

        rows_3ph.append({
            "Timestamp": center_ts,
            "Phase": ph,
            "K1_real": np.real(K1[ph]),
            "K1_imag": np.imag(K1[ph]),
            "K4_real": np.real(K4[ph]),
            "K4_imag": np.imag(K4[ph]),
            "B_S": B[ph],
            "R_Ohm": R[ph],
            "X_Ohm": X[ph],
            "u_R_Ohm": se_R,
            "u_X_Ohm": se_X,
            "T_C": T_est,
            "u_T_C": u_T,
            "rank_step1": res1["rank"],
            "rank_step2": res2["rank"]
        })

results_3ph_long = pd.DataFrame(rows_3ph)

print("\nAntal rader i results_3ph_long:", len(results_3ph_long))
display(results_3ph_long.head())

# ============================================================
# 6. BRETT FORMAT
# ============================================================

out_3ph = results_3ph_long.pivot(
    index="Timestamp",
    columns="Phase",
    values=[
        "K1_real", "K1_imag",
        "K4_real", "K4_imag",
        "B_S", "R_Ohm", "X_Ohm",
        "u_R_Ohm", "u_X_Ohm",
        "T_C", "u_T_C"
    ]
)

out_3ph.columns = [f"{var}_{ph}" for var, ph in out_3ph.columns]
out_3ph = out_3ph.reset_index().sort_values("Timestamp")

print("\nHEAD out_3ph:")
display(out_3ph.head())

# Filtrera bort första fönstret för jämförbarhet
out_3ph_filtered = out_3ph.iloc[1:].copy()

print("\nHEAD out_3ph_filtered:")
display(out_3ph_filtered.head())

# ============================================================
# 7. SUMMARY
# ============================================================

summary_3ph_rows = []

for ph in ["L1", "L2", "L3"]:
    summary_3ph_rows.append({
        "Phase": ph,
        "R_mean_Ohm": out_3ph_filtered[f"R_Ohm_{ph}"].mean(),
        "R_std_Ohm": out_3ph_filtered[f"R_Ohm_{ph}"].std(),
        "R_min_Ohm": out_3ph_filtered[f"R_Ohm_{ph}"].min(),
        "R_max_Ohm": out_3ph_filtered[f"R_Ohm_{ph}"].max(),
        "X_mean_Ohm": out_3ph_filtered[f"X_Ohm_{ph}"].mean(),
        "X_std_Ohm": out_3ph_filtered[f"X_Ohm_{ph}"].std(),
        "T_mean_C": out_3ph_filtered[f"T_C_{ph}"].mean(),
        "T_std_C": out_3ph_filtered[f"T_C_{ph}"].std(),
        "K1_real_mean": out_3ph_filtered[f"K1_real_{ph}"].mean(),
        "K1_imag_mean": out_3ph_filtered[f"K1_imag_{ph}"].mean(),
        "K4_real_mean": out_3ph_filtered[f"K4_real_{ph}"].mean(),
        "K4_imag_mean": out_3ph_filtered[f"K4_imag_{ph}"].mean(),
    })

summary_3ph_df = pd.DataFrame(summary_3ph_rows)

print("\nSUMMARY 3-PH SIMULTAN ROWWISE:")
display(summary_3ph_df)

# ============================================================
# 8. JÄMFÖRELSE MOT TEST-REFERENS
# ============================================================

R_ref_test = 1.98

comparison_3ph_rows = []

for ph in ["L1", "L2", "L3"]:
    R_mean = out_3ph_filtered[f"R_Ohm_{ph}"].mean()
    abs_err = R_mean - R_ref_test
    rel_err_pct = 100 * abs_err / R_ref_test

    comparison_3ph_rows.append({
        "Phase": ph,
        "R_mean_Ohm": R_mean,
        "R_ref_test_Ohm": R_ref_test,
        "Absolute_error_Ohm": abs_err,
        "Relative_error_percent": rel_err_pct
    })

comparison_3ph_df = pd.DataFrame(comparison_3ph_rows)

print("\nJÄMFÖRELSE MOT R_ref_test:")
display(comparison_3ph_df)

# ============================================================
# 9. PLOTTAR
# ============================================================

plt.figure(figsize=(12, 5))
for ph in ["L1", "L2", "L3"]:
    plt.plot(out_3ph_filtered["Timestamp"], out_3ph_filtered[f"R_Ohm_{ph}"], label=ph)
plt.axhline(0, color="black", linestyle="--")
plt.title("Full trefas simultan Singh – resistans per fas – radvis parning")
plt.xlabel("Tid")
plt.ylabel("Resistans (Ohm)")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig(output_dir / "plot_R_3ph_rowwise.png", dpi=200)
plt.show()

plt.figure(figsize=(12, 5))
for ph in ["L1", "L2", "L3"]:
    plt.plot(out_3ph_filtered["Timestamp"], out_3ph_filtered[f"T_C_{ph}"], label=ph)
plt.title("Full trefas simultan Singh – temperatur per fas – radvis parning")
plt.xlabel("Tid")
plt.ylabel("Temperatur (°C)")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig(output_dir / "plot_T_3ph_rowwise.png", dpi=200)
plt.show()

fig, axes = plt.subplots(2, 2, figsize=(14, 8), sharex=True)

for ph in ["L1", "L2", "L3"]:
    axes[0, 0].plot(out_3ph_filtered["Timestamp"], out_3ph_filtered[f"K1_real_{ph}"], label=ph)
    axes[0, 1].plot(out_3ph_filtered["Timestamp"], out_3ph_filtered[f"K1_imag_{ph}"], label=ph)
    axes[1, 0].plot(out_3ph_filtered["Timestamp"], out_3ph_filtered[f"K4_real_{ph}"], label=ph)
    axes[1, 1].plot(out_3ph_filtered["Timestamp"], out_3ph_filtered[f"K4_imag_{ph}"], label=ph)

axes[0, 0].set_title("Re(K1) simultan 3-fas")
axes[0, 1].set_title("Im(K1) simultan 3-fas")
axes[1, 0].set_title("Re(K4) simultan 3-fas")
axes[1, 1].set_title("Im(K4) simultan 3-fas")

for ax in axes.ravel():
    ax.grid(True)
    ax.legend()

plt.tight_layout()
plt.savefig(output_dir / "plot_K1_K4_3ph_rowwise.png", dpi=200)
plt.show()

# ============================================================
# 10. SKILLNADER MELLAN SEND OCH REC
# ============================================================

diff_df = pd.DataFrame(index=df_clean.index)

for ph in ["L1", "L2", "L3"]:
    U_s_mag = df_clean[f"U{ph}MAG_S"]
    U_s_ang = df_clean[f"U{ph}ANG_S"]
    U_r_mag = df_clean[f"U{ph}MAG_R"]
    U_r_ang = df_clean[f"U{ph}ANG_R"]

    I_s_mag = df_clean[f"I{ph}MAG_S"]
    I_s_ang = df_clean[f"I{ph}ANG_S"]
    I_r_mag = df_clean[f"I{ph}MAG_R"]
    I_r_ang = df_clean[f"I{ph}ANG_R"]

    diff_df[f"Delta_U_{ph}_abs"] = U_s_mag - U_r_mag
    diff_df[f"Delta_U_{ph}_pct"] = 100.0 * diff_df[f"Delta_U_{ph}_abs"] / U_r_mag.replace({0: np.nan})

    diff_df[f"Delta_I_{ph}_abs"] = I_s_mag - I_r_mag
    diff_df[f"Delta_I_{ph}_pct"] = 100.0 * diff_df[f"Delta_I_{ph}_abs"] / I_r_mag.replace({0: np.nan})

    diff_df[f"Delta_U_{ph}_deg"] = wrap_angle_deg(U_s_ang - U_r_ang)

    # Korrigerad mottagarströmvinkel med 180 grader
    I_r_ang_corr = wrap_angle_deg(I_r_ang + 180.0)
    diff_df[f"Delta_I_{ph}_deg_corr"] = wrap_angle_deg(I_s_ang - I_r_ang_corr)

    Vs = U_s_mag * np.exp(1j * np.deg2rad(U_s_ang))
    Vr = U_r_mag * np.exp(1j * np.deg2rad(U_r_ang))
    Is_ = I_s_mag * np.exp(1j * np.deg2rad(I_s_ang))
    Ir_corr = I_r_mag * np.exp(1j * np.deg2rad(I_r_ang_corr))

    diff_df[f"P_send_{ph}"] = np.real(Vs * np.conj(Is_))
    diff_df[f"P_rec_{ph}"]  = np.real(Vr * np.conj(Ir_corr))

diff_df["P_total_send"] = diff_df[[f"P_send_{ph}" for ph in ["L1", "L2", "L3"]]].sum(axis=1)
diff_df["P_total_rec"]  = diff_df[[f"P_rec_{ph}" for ph in ["L1", "L2", "L3"]]].sum(axis=1)
diff_df["Delta_P_total"] = diff_df["P_total_send"] - diff_df["P_total_rec"]

print("\nDIFF SUMMARY:")
display(diff_df.describe().T.round(4))

diff_df.to_excel(output_dir / "diff_send_rec_rowwise.xlsx", index=True)

# Plotta vinkelskillnader
fig, axs = plt.subplots(3, 1, sharex=True, figsize=(12, 9))

for i, ph in enumerate(["L1", "L2", "L3"]):
    axs[i].plot(diff_df.index, diff_df[f"Delta_U_{ph}_deg"], label=f"Delta U angle {ph}")
    axs[i].plot(diff_df.index, diff_df[f"Delta_I_{ph}_deg_corr"], label=f"Delta I angle corr {ph}")
    axs[i].axhline(0, color="gray", linewidth=0.6)
    axs[i].set_ylabel("grader")
    axs[i].legend()
    axs[i].grid(True)

axs[-1].set_xlabel("Tid")
plt.suptitle("Vinkeldifferenser med radvis parning")
plt.tight_layout()
plt.savefig(output_dir / "plot_angle_differences_rowwise.png", dpi=200)
plt.show()

# ============================================================
# 11. SPARA RESULTAT
# ============================================================

output_file = output_dir / "singh_3ph_simultaneous_rowwise.xlsx"

with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
    df_joined.reset_index(drop=True).to_excel(writer, sheet_name="df_joined_rowwise", index=False)
    df_clean.reset_index(drop=False).to_excel(writer, sheet_name="df_clean", index=False)
    out_3ph.to_excel(writer, sheet_name="Results_3ph", index=False)
    out_3ph_filtered.to_excel(writer, sheet_name="Results_3ph_filtered", index=False)
    summary_3ph_df.to_excel(writer, sheet_name="Summary_3ph", index=False)
    comparison_3ph_df.to_excel(writer, sheet_name="Compare_Rref", index=False)
    results_3ph_long.to_excel(writer, sheet_name="LongFormat", index=False)
    diff_df.reset_index(drop=False).to_excel(writer, sheet_name="Diff_send_rec", index=False)

print("\nSparat till:", output_file)

# ============================================================
# 12. FRIVILLIG JÄMFÖRELSE MED GAMLA TEST 7-FILEN
# ============================================================

old_file = send_file.parent / "singh_3ph_simultaneous.xlsx"

if old_file.exists():
    try:
        old_summary = pd.read_excel(old_file, sheet_name="Summary_3ph")

        compare_old_new = old_summary.merge(
            summary_3ph_df,
            on="Phase",
            suffixes=("_old_timestamp_join", "_new_rowwise")
        )

        # Skillnader för viktigaste kolumner
        for col in ["R_mean_Ohm", "X_mean_Ohm", "T_mean_C"]:
            old_col = f"{col}_old_timestamp_join"
            new_col = f"{col}_new_rowwise"
            if old_col in compare_old_new.columns and new_col in compare_old_new.columns:
                compare_old_new[f"Delta_{col}_new_minus_old"] = (
                    compare_old_new[new_col] - compare_old_new[old_col]
                )

        print("\nJÄMFÖRELSE GAMMAL TIMESTAMP-JOIN VS NY RADVIS:")
        display(compare_old_new)

        compare_old_new.to_excel(output_dir / "compare_old_timestamp_join_vs_new_rowwise.xlsx", index=False)

    except Exception as e:
        print("\nKunde inte jämföra mot gamla filen.")
        print("Fel:", e)
else:
    print("\nIngen gammal fil hittades för automatisk jämförelse:")
    print(old_file)

print("\nKLART.")
print("Alla resultat finns i:")
print(output_dir)
```
